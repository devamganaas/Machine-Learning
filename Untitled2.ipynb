```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
url="https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
data=pd.read_csv(url)
```

```python
data=data[['mpg','displacement']]
data.dropna(inplace=True)
```

```python
print(data.head)
```

```python
X_train,X_test,y_train,y_test=train_test_split(data['displacement'],data['mpg'],test_size=0.2,random_state=42)
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)
```

```python
print("Linear Regression")
```

```python
print("Linear Regression")
print("MSE:",mean_squared_error(y_test,y_pred_linear))
print("R2 score:",r2_score(y_test,y_pred_linear))
```

```python
degrees=[2,3,4]
for degree in degrees:
    poly_features=PolynomialFeatures(degree=degree)
    X_train_poly=poly_features.fit_transform(X_train.values.reshape(-1, 1))
    X_test_poly=poly_features.transform(X_test.values.reshape(-1, 1))
    model=LinearRegression()
    model.fit(X_train_poly,y_train)
    y_pred=model.predict(X_test_poly)

    print(f"\nPolynomial Regression (Degree {degree})")
    print(f"MSE: {mean_squared_error(y_test, y_pred):.2f}")
    print(f"R2 score: {r2_score(y_test, y_pred):.2f}")
```
